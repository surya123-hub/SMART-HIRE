```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_jobs(resume_text, top_n=10):

    # -----------------------------
    # Step 1: Predict Resume Category
    # -----------------------------
    resume_vector = resume_vectorizer.transform([resume_text])

    pred = resume_model.predict(resume_vector)

    predicted_category = label_encoder.inverse_transform(pred)[0]

    print("Predicted Category:", predicted_category)

    # -----------------------------
    # Step 2: Check if category exists
    # -----------------------------
    if predicted_category not in category_to_jobs:

        print("No jobs available for this category.")

        return None

    # -----------------------------
    # Step 3: Filter Jobs
    # -----------------------------
    filtered_jobs = jobs_df[
        jobs_df["Job_Title"].isin(category_to_jobs[predicted_category])
    ].copy()

    # -----------------------------
    # Step 4: Transform Jobs
    # -----------------------------
    filtered_job_vectors = job_vectorizer.transform(
        filtered_jobs["Search_Text"]
    )

    # -----------------------------
    # Step 5: Transform Resume
    # -----------------------------
    resume_job_vector = job_vectorizer.transform([resume_text])

    # -----------------------------
    # Step 6: Cosine Similarity
    # -----------------------------
    similarity = cosine_similarity(
        resume_job_vector,
        filtered_job_vectors
    ).flatten()

    # -----------------------------
    # Step 7: Attach Similarity
    # -----------------------------
    filtered_jobs["Similarity"] = similarity

    filtered_jobs["Match %"] = (
        filtered_jobs["Similarity"] * 100
    ).round(2)

    # -----------------------------
    # Step 8: Sort
    # -----------------------------
    recommended_jobs = filtered_jobs.sort_values(
        by="Similarity",
        ascending=False
    )

    # -----------------------------
    # Step 9: Return Top Jobs
    # -----------------------------
    return recommended_jobs[
        [
            "Job_Title",
            "Company",
            "City",
            "Experience_Level",
            "Education_Required",
            "Match %"
        ]
    ].head(top_n)
```
