```python
skill_keywords = {

    "Python",
    "Java",
    "C++",
    "C",
    "SQL",
    "MySQL",
    "PostgreSQL",
    "MongoDB",

    "Machine Learning",
    "Deep Learning",
    "TensorFlow",
    "PyTorch",
    "Keras",

    "NLP",
    "Computer Vision",

    "Docker",
    "Kubernetes",

    "AWS",
    "Azure",
    "GCP",

    "Linux",
    "Git",
    "GitHub",

    "JavaScript",
    "React",
    "Node.js",

    "HTML",
    "CSS",

    "Power BI",
    "Tableau",

    "Pandas",
    "NumPy",
    "Scikit-learn",

    "Spark",
    "Hadoop",

    "DevOps",

    "Cybersecurity",

    "Networking",

    "Data Analysis",

    "Excel",

    "REST API"

}
```

```python
def extract_skills(text):

    text = str(text).lower()

    found_skills = []

    for skill in skill_keywords:

        if skill.lower() in text:

            found_skills.append(skill)

    return sorted(list(set(found_skills)))
```

```python
sample_resume = resume_df.loc[1800, "Clean_Resume"]
print(sample_resume)
```

```python
resume_skills = extract_skills(sample_resume)

print(resume_skills)
```

```python
extra_jobs_df["Extracted_Skills"] = extra_jobs_df[
    "Clean_Description"
].apply(extract_skills)
```

```python
recommended_job = "QA Engineer"

mapped_job = job_title_mapping.get(
    recommended_job,
    recommended_job
)

print("Mapped Job:", mapped_job)

matching_job = extra_jobs_df[
    extra_jobs_df["title"].str.contains(
        recommended_job,
        case=False,
        na=False
    )
]
```

```python
required_skills = matching_job.iloc[0]["Extracted_Skills"]

print(required_skills)
```

```python
candidate_skills = set(resume_skills)

required_skills = set(required_skills)

matched_skills = candidate_skills.intersection(required_skills)

missing_skills = required_skills - candidate_skills
```

```python
print("Candidate Skills")

print(sorted(candidate_skills))

print()

print("Required Skills")

print(sorted(required_skills))

print()

print("Matched Skills")

print(sorted(matched_skills))

print()

print("Missing Skills")

print(sorted(missing_skills))
```

```python
if len(required_skills) == 0:

    match_percentage = 0

else:

    match_percentage = (
        len(matched_skills)
        /
        len(required_skills)
    ) * 100

print("Skill Match:", round(match_percentage,2),"%")
```
